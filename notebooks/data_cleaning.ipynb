# 4- Data Cleaning

# (1) Date Conversion

```python
# date conversion for orders table
date_columns = ['order_purchase_timestamp', 'order_approved_at', 
                'order_delivered_carrier_date', 'order_delivered_customer_date', 
                'order_estimated_delivery_date']
for col in date_columns:
    orders[col] = pd.to_datetime(orders[col])
```

```python
orders.info()
```

```python
# date conversion for order_reviews table
order_reviews['review_creation_date'] = pd.to_datetime(order_reviews['review_creation_date'])
order_reviews['review_answer_timestamp'] = pd.to_datetime(order_reviews['review_answer_timestamp'])
```

```python
order_reviews.info()
```

```python
# date conversion for order_items table
order_items['shipping_limit_date'] = pd.to_datetime(order_items['shipping_limit_date'])
```

```python
order_items.info()
```

# (2) Handling Nulls

```python
# In the products table, we found 610 missing values in the names. Instead of deleting them, 
# we will replace them with the word 'Unknown' to ensure that the associated sales data is not lost.

products['product_category_name'] = products['product_category_name'].fillna('Unknown')
```

```python
# Replace missing numerical values (such as weight and dimensions) with median.

cols_to_fix = ['product_weight_g', 'product_length_cm', 'product_height_cm', 'product_width_cm']
for col in cols_to_fix:
    products[col] = products[col].fillna(products[col].median())
```

```python
products.info()
```

Key Takeaways for Cleaning Nulls Phase:
--------------------------------------

- The "Big Clean": The Geolocation table is the only one with a significant duplicate issue that needs to be addressed programmatically to reduce the data size.

- Missing Metadata: We will need to replace the 610 missing values ​​in the Products table with the keyword 'Unknown' before merging and compiling.

- Logical Nulls: The missing values ​​in the Orders table are not data "errors," but rather represent the order journey (orders that have not yet been delivered). Therefore, we will handle them carefully when calculating delivery times.

- Textual Nulls: In the Reviews table, we will focus on the review_score as quantitative data and treat the missing comments as empty messages that do not affect the accuracy of the numerical rating.


# (3) Removing Duplicates

```python
# The geolocation table contains over 260,000 duplicate rows. We will only keep the unique version for each postal code.
# Delete all duplicate rows.

geolocation = geolocation.drop_duplicates()
```

```python
# To ensure one copy for each postal code.

geolocation = geolocation.drop_duplicates(subset=['geolocation_zip_code_prefix'], keep='first')
```

```python
geolocation.info()
```

# (4) Unifying and correcting textual data (Merging Translation)

```python
# We will merge the translation table with the products table to get the names in English directly.

products = pd.merge(products, translation, on='product_category_name', how='left')
```

```python
# If we find products that were not in the translation table,
# we put their original name or Unknown.

products['product_category_name_english'] = products['product_category_name_english'].fillna(products['product_category_name'])
```

```python
products.head()
```

# (5) Calculating "Actual Delivery Time"

```python
# To make sure that the columns are in datetime format

orders[['order_purchase_timestamp', 'order_approved_at', 
                'order_delivered_carrier_date', 'order_delivered_customer_date', 
                'order_estimated_delivery_date']].dtypes

```

```python
# Calculate the actual delivery time in days

orders['actual_delivery_days'] = (orders['order_delivered_customer_date'] - orders['order_purchase_timestamp']).dt.days
```

```python
# Calculating the difference between expected and actual delivery
# Estimated vs Actual (If positive then the delivery was ahead of schedule. if negativethen there was a delay)

orders['delivery_diff_days'] = (orders['order_estimated_delivery_date'] 
                                - orders['order_delivered_customer_date']).dt.days
```

```python
# Displaying the new columns

orders[['actual_delivery_days','delivery_diff_days']].head()
```

```python
# Displaying descriptive statistics for new periods

orders[['actual_delivery_days', 'delivery_diff_days']].describe()
```
